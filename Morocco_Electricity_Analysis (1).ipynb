```python
# Task 2: Data cleaning and preparation
```

```python
cols = ["Temperature", "Humidity", "WindSpeed", "PowerConsumption_Zone1", "PowerConsumption_Zone2",
        "PowerConsumption_Zone3"]
Q1 = df[cols].quantile(0.25)
Q3 = df[cols].quantile(0.75)
IQR = Q3-Q1
((df[cols] < Q1 -1.5 * IQR) | (df[cols] > Q3 + 1.5 * IQR))
```

```python
df.shape
```

```python
df["Year"] = df["Datetime"].dt.year
df["Month"] = df["Datetime"].dt.month
df["Day"] = df["Datetime"].dt.day
df["Hour"] = df["Datetime"].dt.hour
df["DayOfWeek"] = df["Datetime"].dt.day_name()
df["Date"] = df["Datetime"].dt.date
```

```python
#Task 5: Charts
```

```python
#Trend over time
```

```python
daily_avg = df.groupby("Date")[zones].mean()
daily_avg.plot(figsize=(12, 5))
plt.title("Daily average consumption by zone, 2017")
plt.xlabel("Date")
plt.ylabel("Average consumption")
plt.savefig("charts/1_daily_trend.png",dpi=200,bbox_inches="tight")
plt.show()
```

```python
#Average by zone
```

```python
sns.barplot(x=zones, y=df[zones].mean().values)
plt.title("Average consumption by zone")
plt.ylabel("Average consumption")
plt.xticks(rotation=15)
plt.savefig("charts/2_average_by_zone.png",dpi=200,bbox_inches="tight")
plt.show()
```

```python
#Hourly pattern
```

```python
df.groupby("Hour")[zones].mean().plot(figsize=(10,5))
plt.title("Average consumption by hour of day")
plt.xlabel("Hour")
plt.ylabel("Average consumption")
plt.savefig("charts/3_hourly_pattern.png",dpi=200,bbox_inches="tight")
plt.show()
```

```python
#Monthly comparison
```

```python
df. groupby("Month")[zones].mean().plot(kind="bar", figsize=(11,5))
plt.title("Average consumption by month")
plt.xlabel("Month")
plt.ylabel("Average consumption")
plt.savefig("charts/4_monthly_pattern.png",dpi=200,bbox_inches="tight")
plt.show()
```

```python
#Day-of-week comparison
```

```python
order =["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
df.groupby("DayOfWeek")[zones].mean().loc[order].plot(kind="bar", figsize=(10,5))
plt.title("Average consumption by day of week")
plt.ylabel("Average consumption")
plt.savefig("charts/5_day_of_week.png",dpi=200,bbox_inches="tight")
plt.show()
```

```python
#Hour by day heatmap (a seaborn chart showing consumption peaks)
```

```python
pivot = df. pivot_table(values=zones[0], index="Hour", columns="DayOfWeek", aggfunc="mean")[order]
sns.heatmap(pivot, cmap="YlOrRd")
plt.title("Zone 1: average consumption by hour and day")
plt.savefig("charts/6_hour_day_heatmap.png",dpi=200,bbox_inches="tight")
plt.show()
```
